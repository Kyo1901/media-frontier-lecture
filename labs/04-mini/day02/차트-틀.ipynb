```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")             # 🔴 이것을 «먼저» — 글꼴을 되돌리기 때문
plt.rcParams["font.family"] = "Malgun Gothic"  # 한글이 □ 로 깨지지 않게
plt.rcParams["axes.unicode_minus"] = False     # 음수 부호가 □ 로 깨지지 않게

IMG = "web/images/"                          # 차트를 저장할 곳
```

```python
df = pd.read_csv("data/여기에-파일이름.csv")
# 1일차 전처리 셀을 여기에 그대로 붙이세요
```

## 차트 1

```python
fig, ax = plt.subplots(figsize=(8, 4.5))      # 🔴 크기는 전부 같게

# 여기에 그리기 — 예) sns.barplot(x=..., y=..., ax=ax)

ax.set_title("제목 = 이 차트의 결론 한 줄")    # ←
ax.set_xlabel("가로축 이름")                  # ←
ax.set_ylabel("세로축 이름 (단위)")            # ←
fig.savefig(IMG + "1-이름.png", dpi=150, bbox_inches="tight")   # ← 파일 이름은 영어·숫자로
```
